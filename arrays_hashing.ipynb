```python
# 2 Sum 
# Hashmap lookup

def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return (seen[complement], i)
        seen[num] = i
    return None

print(two_sum([2, 7, 11, 15], 9))  # Output: (0, 1)
print(two_sum([3, 2, 4], 6))       # Output:
```

```python
# Contains Duplicate 
# Hashset lookup
def contains_duplicate(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return True
        seen.add(num)
    return False

print(contains_duplicate([1, 2, 3, 1]))  # Output: True
```

```python
# Best Time to Buy and Sell Stock
# Single pass tracking min price
def max_profit(prices):
    min_price = float('inf')
    max_profit = 0
    for price in prices:
        # min_price = min(min_price, price)
        if price < min_price:
            min_price = price
        # max_profit = max(max_profit, price - min_price)
        elif price - min_price > max_profit:
            max_profit = price - min_price
    return max_profit

print(max_profit([7, 1, 5, 3, 6, 4]))  # Output: 5
print(max_profit([7, 6, 4, 3, 1]))
```

```python
# Maximum Subarray
# Kadane's algorithm
def max_sub_array(nums):
    max_current = max_global = nums[0]
    for i in range(1, len(nums)):
        max_current = max(nums[i], max_current + nums[i])
        max_global = max(max_global, max_current)
    return max_global

print(max_sub_array([-2,1,-3,4,-1,2,1,-5,4]))  # Output: 6
```

```python
# Missing Number
# Hashset lookup or sum formula
def missing_number(nums):
    n = len(nums)
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(nums)
    return expected_sum - actual_sum

print(missing_number([3, 0, 1]))  # Output: 2
```

```python
# Intersection of Two Arrays
# Hashset lookup
def intersection(nums1, nums2):
    set1 = set(nums1)
    result = set()
    for num in nums2:
        if num in set1:
            result.add(num)
    return list(result)

print(intersection([1, 2, 2, 1], [2, 2]))  # Output: [2]
```

```python
# Product of Array Except Self
# Prefix and suffix products
def product_except_self(nums):
    n = len(nums)
    output = [1] * n
    prefix = 1
    for i in range(n):
        output[i] = prefix
        prefix *= nums[i]
    suffix = 1
    for i in range(n - 1, -1, -1):
        output[i] *= suffix
        suffix *= nums[i]
    return output

print(product_except_self([1, 2, 3, 4]))  # Output: [24, 12, 8, 6]
```

```python
# Top K Frequent Elements
# Hashmap + bucket sort
def top_k_frequent(nums, k):
    frequency = {}
    for num in nums:
        frequency[num] = frequency.get(num, 0) + 1
    
    sorted_freq = sorted(frequency.items(), key=lambda x: x[1], reverse=True)
    return [num for num, freq in sorted_freq[:k]]

print(top_k_frequent([1, 1, 1, 2, 2, 3], 2))  # Output: [1, 2]
```

```python
# Longest Consecutive Sequence
# Hashset lookup
def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0
    for num in num_set:
        if num - 1 not in num_set:  # Start of a sequence
            length = 1
            while num + length in num_set:
                length += 1
            longest = max(longest, length)
    return longest

print(longest_consecutive([100, 4, 200, 1, 3, 2]))  # Output: 4

def longestConsecutive(nums):
        # write your code here
        nums.sort()
        global_maximum = 1
        maximum = 1
        for i in range(len(nums) -1):
            if nums[i+1] == nums[i] + 1:
                maximum += 1
                global_maximum = max(global_maximum, maximum)
            else:
                maximum = 1
        return global_maximum

print(longestConsecutive([100, 4, 200, 1, 3, 2]))  # Output: 4
```

```python
# Subarray Sum Equals K
# Hashmap of prefix sums
def subarray_sum(nums, k):
    count = 0
    prefix_sum = 0
    prefix_sums = {0: 1}  # To handle the case when prefix_sum == k
    for num in nums:
        prefix_sum += num
        if prefix_sum - k in prefix_sums:
            count += prefix_sums[prefix_sum - k]
        prefix_sums[prefix_sum] = prefix_sums.get(prefix_sum, 0) + 1
    return count

print(subarray_sum([1, 4, 7, 8, 10], 11))  # Output: 1
print(subarray_sum([1, 2, 3], 3))  # Output: 2
```

```python
# Find Duplicate Number
# Hashset lookup
def find_duplicate(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return num
        seen.add(num)
    return None

print(find_duplicate([1, 3, 4, 2, 2, 3]))  # Output: 2
```

```python
# Merge Intervals
# Sort and merge
def merge_intervals(intervals):
    if not intervals:
        return []
    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]
    for current in intervals[1:]:
        last = merged[-1]
        if current[0] <= last[1]:  # Overlap
            last[1] = max(last[1], current[1])  # Merge
        else:
            merged.append(current)
    return merged

print(merge_intervals([[1, 3], [2, 6], [8, 10], [15, 18]]))  # Output: [[1, 6], [8, 10], [15, 18]]
```

```python
# Majority Element
# Hashmap counting or Boyer-Moore Voting Algorithm
def majority_element(nums):
    count = 0
    candidate = None
    for num in nums:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)
    return candidate
print(majority_element([2, 2, 1, 1, 1, 2, 2]))  # Output: 2

def majorityElement(nums):
    freq = {}
    for num in nums:
        freq[num] = freq.get(num, 0) + 1
    majority = max(freq, key=freq.get)
    return majority
print(majorityElement([2, 2, 1, 1, 1, 2, 2]))  # Output: 2
```

```python
# Find repeating and missing number
# Hashmap counting
def find_repeating_and_missing(nums):
    count = {}
    for num in nums:
        count[num] = count.get(num, 0) + 1
    
    repeating = missing = None
    for i in range(1, len(nums) + 1):
        if i in count:
            if count[i] > 1:
                repeating = i
        else:
            missing = i
    return repeating, missing
print(find_repeating_and_missing([3, 1, 2, 5, 3]))  # Output: (3, 4)
```

```python
# Check if array is sorted(rotate)
# Single pass check
def is_sorted(nums):
    count = 0
    for i in range(0, len(nums)):
        if nums[i] > nums[(i + 1) % len(nums)]:
            count += 1

    return count <= 1
print(is_sorted([3, 4, 5, 1, 2]))  # Output: True
print(is_sorted([1, 2, 3, 4, 5]))  # Output: True
```

```python
# Valid sudoku
# Hashset lookup for rows, columns, and boxes
def is_valid_sudoku(board):
    for i in range(9):
        row = set()
        col = set()
        box = set()
        for j in range(9):
            # Check row
            if board[i][j] != '.':
                if board[i][j] in row:
                    return False
                row.add(board[i][j])
            # Check column
            if board[j][i] != '.':
                if board[j][i] in col:
                    return False
                col.add(board[j][i])
            # Check box
            box_row = 3 * (i // 3) + j // 3
            box_col = 3 * (i % 3) + j % 3
            if board[box_row][box_col] != '.':
                if board[box_row][box_col] in box:
                    return False
                box.add(board[box_row][box_col])
    return True

sudoku_board = [
    ["5","3",".",".","7",".",".",".","."],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]
print(is_valid_sudoku(sudoku_board))  # Output: True
```

```python
# Set Matrix Zeroes

def setZeroes(matrix):
    rows, cols = len(matrix), len(matrix[0])
    zero_rows, zero_cols = set(), set()

    # Identify positions of zeros
    for i in range(rows):
        for j in range(cols):
            if matrix[i][j] == 0:
                zero_rows.add(i)
                zero_cols.add(j)

    # Set entire rows and columns to zero
    for i in range(rows):
        for j in range(cols):
            if i in zero_rows or j in zero_cols:
                matrix[i][j] = 0

    return matrix

print(setZeroes([[1, 1, 1], [1, 0, 1], [1, 1, 1]]))  # Output: [[1, 0, 1], [0, 0, 0], [1, 0, 1]]
```

```python
# rotate array
def rotate_array(nums, k):
    n = len(nums)
    k %= n  # In case k is greater than n
    nums[:] = nums[-k:] + nums[:-k]  # Rotate in place
    return nums

print(rotate_array([1, 2, 3, 4, 5, 6, 7], 3))  # Output: [5, 6, 7, 1, 2, 3, 4]

def rotate_array_in_place(nums, k):
    n = len(nums)
    k %= n  # In case k is greater than n
    def reverse(start, end):
        while start < end:
            nums[start], nums[end] = nums[end], nums[start]
            start += 1
            end -= 1
    reverse(0, n - 1)
    reverse(0, k - 1)
    reverse(k, n - 1)
    return nums

print(rotate_array_in_place([1, 2, 3, 4, 5, 6, 7], 3))  # Output: [5, 6, 7, 1, 2, 3, 4]
```

```python
# Rotate Image
# Transpose and reverse
def rotate_image(matrix):
    n = len(matrix)
    # Transpose the matrix
    for i in range(n):
        for j in range(i, n):
            matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
    # Reverse each row
    for i in range(n):
        matrix[i].reverse()
    return matrix

print(rotate_image([[1,2,3],[4,5,6],[7,8,9]]))  # Output: [[7,4,1],[8,5,2],[9,6,3]]
```
