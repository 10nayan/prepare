```python
import time
import threading
from concurrent.futures import ThreadPoolExecutor, Future

def send_sms():
    print("Sending SMS...")
    time.sleep(2)  # Simulate time taken to send an SMS
    print("SMS sent!")


def send_email():
    print("Sending email...")
    time.sleep(3)  # Simulate time taken to send an email
    print("Email sent!")

def calculate_eta():
    print("Calculating ETA...")
    time.sleep(4)  # Simulate time taken to calculate ETA
    return "ETA: 15 minutes"

if __name__ == "__main__":
    with ThreadPoolExecutor(max_workers=3) as executor:
        future_sms = executor.submit(send_sms)
        future_email = executor.submit(send_email)
        future_eta = executor.submit(calculate_eta)

        print("Starting tasks...")
        future_sms.result()  # Wait for SMS to complete
        future_email.result()  # Wait for email to complete
        eta_result = future_eta.result()  # Wait for ETA calculation to complete
        print(eta_result)
        print("All tasks completed!")
```

```python
import time
import threading
from concurrent.futures import ThreadPoolExecutor, Future

def send_sms():
    print("Sending SMS...")
    time.sleep(2)  # Simulate time taken to send an SMS
    print("SMS sent!")


def send_email():
    print("Sending email...")
    time.sleep(3)  # Simulate time taken to send an email
    print("Email sent!")

def calculate_eta():
    print("Calculating ETA...")
    time.sleep(4)  # Simulate time taken to calculate ETA
    return "ETA: 15 minutes"

if __name__ == "__main__":
    with ThreadPoolExecutor(max_workers=3) as executor:
        futures = {
            "sms": executor.submit(send_sms),
            "email": executor.submit(send_email),
            "eta": executor.submit(calculate_eta)
        }
        print("Starting tasks...")
        for name, future in futures.items():
            future.result()  # Wait for each task to complete
            if name == "eta":
                print(future.result())  # Print ETA result
        print("All tasks completed!")
```
