1. Find a pair whose sum equals a given number

Time: O(n) using hashing.

```python
# Hashing
def find_pair(arr, target):
    seen = set()
    for num in arr:
        if target - num in seen:
            return (target - num, num)
        seen.add(num)
    return None

def find_pair_index(arr, target):
    seen = {}
    for i, num in enumerate(arr):
        if target - num in seen:
            return (seen[target - num], i)
        seen[num] = i
    return None

# Test
print(find_pair([2,7,11,15], 9))
print(find_pair_index([2,7,11,15], 9))
```

2. Find all pairs whose sum equals a given number

Time: O(n)

```python
# Hashing
def all_pairs(arr, target):
    seen = set()
    pairs = []

    for num in arr:
        if target - num in seen:
            pairs.append((target - num, num))
        seen.add(num)

    return pairs

# Test
print(all_pairs([1,2,3,4,5,6], 7))
```

3. Find max and min without built-in functions

```python
def find_min_max(arr):
    min_val = arr[0]
    max_val = arr[0]

    for i in range(1, len(arr)):
        if arr[i] < min_val:
            min_val = arr[i]
        if arr[i] > max_val:
            max_val = arr[i]

    return min_val, max_val

# Test
print(find_min_max([3,7,1,9,4]))
```

4. Remove duplicates from a list

```python
def remove_duplicates(arr):
    seen = {}
    result = []

    for i in range(len(arr)):
        if arr[i] not in seen:
            seen[arr[i]] = True
            result.append(arr[i])

    return result

# Test
print(remove_duplicates([1,2,2,3,4,4,5]))
```

5. Find second largest number

```python
def second_largest(arr):
    first = float('-inf')
    second = float('-inf')

    for i in range(len(arr)):
        if arr[i] > first:
            second = first
            first = arr[i]
        elif arr[i] > second and arr[i] != first:
            second = arr[i]

    return second

# Test
print(second_largest([10,5,20,8,15]))
```

6. Rotate list left by k

```python
def rotate_left(arr, k):
    n = len(arr)
    k = k % n

    result = []

    for i in range(k, n):
        result.append(arr[i])

    for i in range(k):
        result.append(arr[i])

    return result

# Test
print(rotate_left([1,2,3,4,5],2))
```

7. Rotate list right by k

```python
def rotate_right(arr, k):
    n = len(arr)
    k = k % n

    result = []

    for i in range(n-k, n):
        result.append(arr[i])

    for i in range(n-k):
        result.append(arr[i])

    return result

# Test
print(rotate_right([1,2,3,4,5],2))
```

8. Merge two sorted lists

```python
def merge_sorted(a, b):
    i = 0
    j = 0
    result = []

    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    while i < len(a):
        result.append(a[i])
        i += 1

    while j < len(b):
        result.append(b[j])
        j += 1

    return result

# Test
print(merge_sorted([1,3,5], [2,4,6]))
```

9. Find intersection of two lists

```python
def intersection(arr1, arr2):
    seen = {}
    result = []

    for i in range(len(arr1)):
        seen[arr1[i]] = True

    for i in range(len(arr2)):
        if arr2[i] in seen:
            result.append(arr2[i])

    return result

# Test
print(intersection([1,2,3,4], [3,4,5,6]))
```

10. Find missing number (1 to n)

```python
def missing_number(arr, n):
    total = n * (n + 1) // 2

    return total - sum(arr)

# Test
print(missing_number([1,2,4,5],5))
```

11. Move all the zeros to the last


```python
# Two pointers
def moveZeroes(nums):
    k = 0
    for i in range(0, len(nums)):
        if nums[i] != 0:
            nums[k], nums[i] = nums[i], nums[k]
            k += 1
            
    return nums
moveZeroes([1,2,0,5,7,0])
```

12. Best Time to Buy and Sell Stock

```python

def maxProfit(prices):
    minimum = float('inf')
    profit = 0
    for price in prices:
        minimum = min(minimum, price)
        profit = max(profit, price - minimum)
    return profit

print(maxProfit([8,7,1,1,6,9]))
```

13. Find maximum subarray

```python
# Greedy/DP
def max_sub_arr(nums):
    max_cur = global_max = nums[0]
    for i in range(1, len(nums)):
        max_cur = max(max_cur, max_cur + nums[i])
        if max_cur > global_max:
            global_max = max_cur
    return global_max

print(max_sub_arr([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
```

15. Product of Array Except Self

```python
# Prefix suffix
def productExceptSelf(nums):
    n = len(nums)
    res = [1]*n

    prefix = 1
    for i in range(n):
        res[i] = prefix
        prefix *= nums[i]

    suffix = 1
    for i in range(n-1, -1, -1):
        res[i] *= suffix
        suffix *= nums[i]

    return res

print(productExceptSelf([1,2,3,4]))
```

16. Container With Most Water

```python
# Two pointer
def max_area(height):
    max_area = 0
    left = 0
    right = len(height) - 1
    while left < right:
        width = right - left
        area = width * min(height[left], height[right])
        max_area = max(max_area, area)
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    return max_area

print(max_area([1,8,6,2,5,4,8,3,7]))
```

17. 3 Sum
Find all triplets that sum to zero.

```python
# Two pointer
def threeSum(nums):
    res = []
    nums.sort()  # Step 1: Sort

    for i in range(len(nums)):
        # Skip the same value for 'a' to avoid duplicate triplets
        if i > 0 and nums[i] == nums[i-1]:
            continue
            
        # Two-pointer search for 'b' and 'c'
        left, right = i + 1, len(nums) - 1
        while left < right:
            three_sum = nums[i] + nums[left] + nums[right]
            
            if three_sum > 0:
                right -= 1
            elif three_sum < 0:
                left += 1
            else:
                res.append([nums[i], nums[left], nums[right]])
                left += 1
                
                # Skip duplicate values for 'b'
                while nums[left] == nums[left-1] and left < right:
                    left += 1
                    
    return res

print(threeSum([-1,0,1,2,-1,-4]))
```

18. Subarray Sum Equals K

```python
# Hashing
def subarray_sum(nums, k):
    count = 0
    current_sum = 0
    prefixes = {0:1}

    for i in nums:
        current_sum += i
        diff = current_sum - k
        if diff in prefixes:
            count += prefixes[diff]
        
        prefixes[current_sum] = prefixes.get(current_sum, 0) + 1
    return count

print(subarray_sum([1,1,1], 2))
```

19. Longest Substring Without Repeating Characters

```python
# Set
def length_of_longest_substring(s):
    char_map = {}  # Stores {character: last_seen_index}
    max_length = 0
    left = 0

    for right in range(len(s)):
        char = s[right]
        
        # If we've seen the character and it's inside our current window
        if char in char_map and char_map[char] >= left:
            # Jump the left pointer to the right of the previous occurrence
            left = char_map[char] + 1
        
        # Update the last seen position of the character
        char_map[char] = right
        
        # Calculate window size and update max
        max_length = max(max_length, right - left + 1)

    return max_length
print(length_of_longest_substring("abcabcbb"))

```
