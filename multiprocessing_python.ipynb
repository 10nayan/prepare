```python
# This code demonstrates the use of the multiprocessing module in Python to create and manage multiple processes. Each process runs a worker function that prints its process ID. The main block creates and starts several processes, then waits for them to complete before printing a final message.
# NOTE: This will not run in a Jupyter notebook environment due to the way multiprocessing works. It should be run as a standalone Python script.

import multiprocessing
import os

def worker(num):
    print(f"Worker {num} is running in process ID: {os.getpid()}")

if __name__ == "__main__":
    processes = []
    for i in range(5):
        p = multiprocessing.Process(target=worker, args=(i,))
        processes.append(p)
        p.start()

    for p in processes:
        p.join()

    print("All processes completed!")
```
